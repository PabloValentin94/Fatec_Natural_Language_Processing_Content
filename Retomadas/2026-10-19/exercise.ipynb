### 02 - Recursos (Importações)

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import pandas as pd
```

### 03 - Implementação

```python
corpus = [
  "machine learning é usado para previsão de vendas",
  "a bolsa de valores registrou alta no mercado financeiro",
  "análise de dados ajuda empresas a tomar decisões",
  "algoritmos de aprendizado de máquina analisam grandes volumes de dados",
]

vectorizer = TfidfVectorizer()

matrix = vectorizer.fit_transform(corpus)

print(
  "Vocabulário:",
  "\n\n",
  vectorizer.get_feature_names_out()
)

print(
  "\nMatriz:",
  "\n\n",
  matrix.toarray()
)

print(
  "\nDataFrame:",
  "\n\n",
  pd.DataFrame(
    data=matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=[f"Documento {i + 1}" for i in range(0, len(corpus))]
  )
)
```

### 04 - Preparação da Consulta

```python
query = "aprendizado de máquina e dados"

query_matrix = vectorizer.transform([query])

print(
  "Matriz:",
  "\n\n",
  query_matrix.toarray()
)

print(
  "\nDataFrame:",
  "\n\n",
  pd.DataFrame(
    data=query_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
  )
)
```

### 05 - Similaridades (Consulta/Documentos)

```python
similarities = cosine_similarity(query_matrix, matrix)

print(
  "Similaridades:",
  "\n\n",
  similarities
)

print(
  "\nDataFrame:",
  "\n\n",
  pd.DataFrame(
    data=similarities,
    columns=[f"Documento {i + 1}" for i in range(0, len(corpus))],
    index=["Consulta"]
  )
)
```

### 06 - Relevância dos Documentos

```python
sorted_documents = [
  (f"Documento {i + 1}", similarities[0][i]) for i in range(0, len(corpus))
]

sorted_documents.sort(
  key=lambda document: document[1],
  reverse=True
)

print(
  f"Consulta:\n\n{query}"
)

print("\nSimilaridade dos Documentos com a Consulta:")

for (document, similarity) in sorted_documents:
  print(
    f"\n| {document:14}",
    f"| {(similarity * 100):8.2f}% |"
  )
```
