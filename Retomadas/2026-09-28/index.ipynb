```python
import spacy
```

### 03 - Implementação

```python
nlp_01 = spacy.load("pt_core_news_sm")

text_01 = "O aluno Pablo Valentin deve gravar um vídeo sobre os conceitos básicos de Processamento de linguagem Natural."

document_01 = nlp_01(text_01)

print("Tokens:\n")

for token in document_01:
  print(
    f"| Texto: {token.text:20}", # Termo (Token).
    f"| Lemma: {token.lemma_:20}", # Forma Base (Token).
    f"| POS: {token.pos_:10}", # Classificação (Token).
    f"| Stop Word: {('Sim' if token.is_stop == 1 else 'Não'):10} ({int(token.is_stop)}) |" # Retorna se é uma Stop Word (Palavra Irrelevante).
  )

print("\nEntidades:\n")

for entity in document_01.ents:
  print(
    f"| Texto: {entity.text:40}",
    f"| Label: {entity.label_:10} |",
  )

print("\nSentenças:\n")

for sentence in document_01.sents:
  print(f"| Sentença: {sentence.text} |") # Nesse caso vai ser apenas uma.
```

### 05 - Dependência Sintática

```python
for token in document_01:
  print(
    f"| Texto: {token.text:20}", # Termo (Token).
    f"| Lemma: {token.lemma_:20}", # Forma Base (Token).
    f"| POS: {token.pos_:10}", # Classificação (Token).
    f"| Termo Relacionado (Token): {token.head.text:20}", # Termo Relacionado (Token).
    f"| Classificação da Relação (DEP): {token.head.dep_:10} |" # Tipo de Relação (DEP).
  )
```

### 06 - Similaridade

```python
nlp_02 = spacy.load("pt_core_news_lg")

text_02_01 = "Inteligência Artificial"

document_02_01 = nlp_02(text_02_01)

text_02_02 = "Apredizado de Máquina"

document_02_02 = nlp_02(text_02_02)

print("Similaridade entre os Documentos:\n")

print(
  f"| Porcentagem: {(document_02_01.similarity(document_02_02) * 100):.2f}%",
  f"| Decimal (Bruto): {document_02_01.similarity(document_02_02)} |"
)
```
