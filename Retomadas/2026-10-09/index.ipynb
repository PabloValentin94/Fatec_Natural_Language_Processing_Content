### 02 - Recursos (Importações)

```python
from sklearn.feature_extraction.text import CountVectorizer

import pandas as pd
```

### 03 - Implementação

```python
corpus = [
  "gosto de machine learning",
  "machine learning é incrível",
  "gosto de programar"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(corpus)

print(
  "Vocabulário:",
  "\n\n",
  f"{vectorizer.get_feature_names_out()}"
)

print(
  "\nMatriz:",
  "\n\n",
  X.toarray()
)

print(
  "\nDataFrame:",
  "\n\n",
  pd.DataFrame(
    data=X.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=[f"Documento {i + 1}" for i in range(0, len(corpus))]
  )
)
```

### 04 - Exemplo de Esparsidade

```python
sparse_corpus = [
  *corpus,
  """
    O conhecimento transforma vidas, inspira novas ideias, fortalece amizades e abre caminhos para construir um futuro melhor, repleto de oportunidades e realizações.
  """.strip()
]

sparse_vectorizer = CountVectorizer()

sparse_X = sparse_vectorizer.fit_transform(sparse_corpus)

print(
  "Vocabulário:",
  "\n\n",
  f"{sparse_vectorizer.get_feature_names_out()}"
)

print(
  "\nMatriz:",
  "\n\n",
  sparse_X.toarray()
)

print(
  "\nDataFrame:",
  "\n\n",
  pd.DataFrame(
    data=sparse_X.toarray(),
    columns=sparse_vectorizer.get_feature_names_out(),
    index=[f"Documento {i + 1}" for i in range(0, len(sparse_corpus))]
  )
)
```
