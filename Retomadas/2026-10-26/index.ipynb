### 02 - Recursos (Importações)

```python
from gensim.models import Word2Vec
```

### 03 - Implementação

```python
corpus = [
  ["gato", "come", "peixe"],
  ["cachorro", "come", "carne"],
  ["gato", "e", "cachorro", "sao", "animais"]
]

model = Word2Vec(
  sentences=corpus,
  vector_size=5,
  window=2,
  min_count=1
)

vocabulary = model.wv.index_to_key

print("Token | Representação Vetorial:")

for token in vocabulary:
  print(f"\n- {token:10} <--->   {model.wv[token]}")
```

### 04 - Similaridade Específica

```python
# Similaridade entre 2 tokens.

example_similarity = model.wv.similarity("gato", "cachorro")

print(f"Similaridade entre 'Gato' e 'Cachorro': {example_similarity:.20}")
```

### 05 - Similaridade Geral

```python
# Saída total.

print(f"Similaridade (Gato)")

for (token, score) in model.wv.most_similar("gato"):
  print(
    f"\n| Palavra: {token:8}",
    f"| Proximidade: {score:.20f} |"
  )

# Saída parcial.

print(f"\nSimilaridade (Cachorro)")

for (token, score) in model.wv.most_similar("cachorro", topn=3): # O segundo parâmetro limita a quantidade de tokens na saída.
  print(
    f"\n| Palavra: {token:8}",
    f"| Proximidade: {score:.20f} |"
  )
```
