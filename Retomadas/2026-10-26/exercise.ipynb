### 02 - Recursos (Importações)

```python
from gensim.models import Word2Vec
```

### 03 - Treinamento

```python
corpus = [
  ["o", "processamento", "de", "linguagem", "natural", "e", "incrivel"],
  ["aprendizado", "de", "maquina", "e", "uma", "subarea", "da", "ia"],
  ["redes", "neurais", "sao", "usadas", "em", "deep", "learning"],
  ["o", "gensim", "e", "excelente", "para", "word", "embeddings"]
]

model = Word2Vec(
  sentences=corpus,
  vector_size=100,
  window=5,
  min_count=1
)

vocabulary = model.wv.index_to_key

print(
  "Vocabulário:",
  "\n\n",
  vocabulary,
)

print("\nToken | Representação Vetorial:")

for token in vocabulary:
  print(f"\n- {token} <---> {model.wv[token]}")
```

### 04 - Similiaridade entre Tokens

```python
target_token = "linguagem"

similar_tokens = model.wv.most_similar(target_token, topn=3)

print(f"Similaridade (Linguagem)")

for (token, score) in similar_tokens:
  print(
    f"\n| Palavra: {token:8}",
    f"| Proximidade: {score:.20f} |"
  )
```
