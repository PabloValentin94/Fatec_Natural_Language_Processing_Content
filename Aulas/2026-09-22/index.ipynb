### 02 - Recursos (Importações)

```python
import sounddevice as sd
import scipy.io.wavfile as wav
import whisper
import pyttsx3
import time
```

### 03 - Funções

```python
def record_sound(filename: str = "sound.wav", duration: int = 5, fs: int = 44100):
  print("\nPreparando...")

  for i in range(3, 0, -1):
    print(f"\n{i}")

    time.sleep(1)

  print("\nGravando Áudio...")

  sound = sd.rec(duration * fs, samplerate=fs, channels=1)

  sd.wait()

  wav.write(filename, fs, sound)

  print("\nGravação Finalizada.")

def transcribe_sound(filename: str = "sound.wav"):
  print("\nTranscrevendo Áudio...")

  model = whisper.load_model("medium")

  result = model.transcribe(filename, language="pt")

  text = result["text"]

  print("\nTranscrição concluída.")

  print(f"\nTexto: {text}")

  return text.lower()

def generate_response(text):
  normalized_text = text.lower()

  if ("oi" in normalized_text):
    return "Olá! Como posso ajudar?" 
  elif ("nome" in normalized_text):
    return "Eu sou um assistente virtual feito em Python"
  elif ("hora" in normalized_text):
    return time.strftime("Agora são %H horas e %M minutos.")
  else:
    return "Não foi possível obter uma resposta."

def synthesize_voice(text):
  print("\nGerando Síntese de Voz...")

  engine = pyttsx3.init()

  engine.setProperty("rate", 180)
  engine.setProperty("volume", 1.0)

  voices = engine.getProperty("voices")

  for voice in voices:
    if ("brazil" in voice.name.lower() or "portuguese" in voice.name.lower()):
      engine.setProperty("voice", voice.id)

  engine.say(text)

  engine.runAndWait()

  print("\nSíntese de Voz Gerada com Sucesso.")

def execute_pipeline():
  record_sound()

  text = transcribe_sound("sound.wav")

  response = generate_response(text)

  print(f"\nResposta: {response}")

  synthesize_voice(response)

def display_voices_options():
  engine = pyttsx3.init()

  voices = engine.getProperty("voices")

  print("Vozes Disponíveis (Idiomas):")

  for voice in voices:
    print(f"\n{voice.name:20}:\n")

    print(f"- Linguagens: {voice.languages}\n",)
    print(f"- Gênero: {voice.gender:20}\n",)
    print(f"- Idade: {voice.age:20}")
```

### 04 - Execução (Pipeline)

```python
execute_pipeline()
```

### 05 - Vozes Disponíveis

```python
display_voices_options()
```
