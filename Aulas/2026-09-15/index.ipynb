### 02 - Recursos (Importações)

```python
import sounddevice as sd
import scipy.io.wavfile as wav
import whisper

from whisper import Whisper
from datetime import datetime
```

### 03 - Funções

```python
def record_sound(filename: str = "sound.wav", duration: int = 5, fs: int = 44100):
  print("\nGravando Áudio...")

  sound = sd.rec(duration * fs, samplerate=fs, channels=1)

  sd.wait()

  wav.write(filename, fs, sound)

  print("\nGravação Finalizada.")

def transcribe_sound(model: Whisper, filename: str = "sound.wav"):
  print("\nTranscrevendo Áudio...")

  result = model.transcribe(filename, language="pt")

  text = result["text"]

  print("\nTranscrição concluída.")

  print(f"\nTexto: {text}")

  return text.lower()

def classify_commands(text):
  if ("luz" in text and "ligar" in text):
    print("\nLigue a luz.")
  elif ("luz" in text and "desligar" in text):
    print("\nDesligue a luz.")
  elif ("hora" in text):
    print("\nHora: ", datetime.now().strftime("%H:%M"))
  elif ("data" in text):
    print("\nData: ", datetime.now().strftime("%d/%m/%Y"))
  elif ("basicamente" in text):
    print("\nEssa palabra não está na Bíblia.")
  else:
    print("\nComando não reconhecido.")

def execute_pipeline(model: Whisper):
  record_sound()

  text = transcribe_sound(model)

  classify_commands(text)

  global sound_text

  sound_text = text
```

### 04 - Modelos Whisper

```python
print("Carregando Modelo (Whisper)...\n")

model = whisper.load_model("small")

print("\nModelo Carregado.")
```

### 05 - Execução (Pipeline)

```python
execute_pipeline(model)
```
