### 02 - Recursos (Importações)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

import pandas as pd
```

### 03 - Resolução

```python
corpus = [
  "O gato dorme no sofá",
  "O cachorro dorme no tapete",
  "O gato caça o rato",
  "O cachorro corre no parque"
]

vectorizer = TfidfVectorizer()

matrix = vectorizer.fit_transform(corpus)

print("Tabela (TF - IDF):\n")

dataframe = pd.DataFrame(
  matrix.toarray(),
  columns=vectorizer.get_feature_names_out(),
  index=["Documento 01", "Documento 02", "Documento 03", "Documento 04"]
)

print(dataframe)
```

### 05 - Palavras Comuns (Stopwords)

No contexto atual, a palavra **"No"** é considerada uma **Stopword** (Artigos, Preposições, etc.), mesmo que não seja removida pelo conversor, pois é um termo que serve apenas como ligação entre outras duas estruturas. Devido a isso, é muito utilizada em várias frases, o que faz com que com que seu IDF seja baixo, ou seja, é classificada como uma palavra comum.
