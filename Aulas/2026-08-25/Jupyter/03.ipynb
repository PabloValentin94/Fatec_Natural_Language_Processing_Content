### 02 - Recursos (Importações)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

import pandas as pd
```

### 03 - Resolução

```python
corpus = [
  "O banco aprovou o empréstimo do cliente",
  "O cliente solicitou um empréstimo no banco",
  "O jogador marcou um gol no campeonato",
  "O time venceu o campeonato de futebol"
]

vectorizer = TfidfVectorizer()

matrix = vectorizer.fit_transform(corpus)

print("Tabela (TF - IDF):\n")

dataframe = pd.DataFrame(
  matrix.toarray(),
  columns=vectorizer.get_feature_names_out(),
  index=["Documento 01", "Documento 02", "Documento 03", "Documento 04"]
)

print(dataframe)
```
