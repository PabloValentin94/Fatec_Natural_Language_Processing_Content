### 02 - Recursos (Importações)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

import pandas as pd
```

### 03 - Resolução

```python
corpus = [
  "Python é muito usado para ciência de dados",
  "Java é muito usado no desenvolvimento corporativo",
  "Python é muito popular em machine learning",
  "JavaScript é muito usado no desenvolvimento web"
]

vectorizer = TfidfVectorizer()

matrix = vectorizer.fit_transform(corpus)

print("Tabela (TF - IDF):\n")

dataframe = pd.DataFrame(
  matrix.toarray(),
  columns=vectorizer.get_feature_names_out(),
  index=["Documento 01", "Documento 02", "Documento 03", "Documento 04"]
)

print(dataframe)

for (document, row) in dataframe.iterrows():
  print(f"\nPalavras com mais Peso ({document}):\n")

  for (term, tfidf) in row.sort_values(ascending=False).head(3).items():
    print(
      f"| Palavra: {term:20}",
      f"| TF - IDF: {tfidf:20} |"
    )
```
