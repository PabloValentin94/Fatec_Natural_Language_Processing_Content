### 02 - Recursos (Importações)

```python
from sklearn.feature_extraction.text import CountVectorizer
```

### 03 - Comparação Rasa

```python
corpus_01 = [
  "Eu gosto deste filme, não é ruim."
]

vectorizer_01 = CountVectorizer()

X_01 = vectorizer_01.fit_transform(corpus_01)

words_01 = vectorizer_01.get_feature_names_out()

print(f"Vocabulário (1ª Frase):\n\n{words_01}")

corpus_02 = [
  "Eu não gosto deste filme, é ruim."
]

vectorizer_02 = CountVectorizer()

X_02 = vectorizer_02.fit_transform(corpus_02)

words_02 = vectorizer_02.get_feature_names_out()

print(f"\nVocabulário (2ª Frase):\n\n{words_02}")

print("\nAmbos são idênticos.")
```

### 04 - Comparação Profunda

```python
corpus_03 = [
  "Eu gosto deste filme, não é ruim."
]

vectorizer_03 = CountVectorizer(ngram_range=(1,2))

X_03 = vectorizer_03.fit_transform(corpus_03)

words_03 = vectorizer_03.get_feature_names_out()

print(f"Vocabulário (3ª Frase):\n\n{words_03}")

corpus_04 = [
  "Eu não gosto deste filme, é ruim."
]

vectorizer_04 = CountVectorizer(ngram_range=(1,2))

X_04 = vectorizer_04.fit_transform(corpus_04)

words_04 = vectorizer_04.get_feature_names_out()

print(f"\nVocabulário (4ª Frase):\n\n{words_04}")

print("\nHá leves diferenças na estrutura do Bag of Words, apesar de ambas as frases usarem as mesmas palavras.")
```
