### 02 - Recursos (Importações)

```python
from sklearn.feature_extraction.text import CountVectorizer
```

### 03 - Resolução

```python
corpus = [
  "Ganhe um prêmio incrível agora! Clique no link para participar.", # Spam.
  "Parabéns! Você foi selecionado para receber R$ 1.000. Acesse agora!", # Spam.
  "Oferta exclusiva! Compre hoje e ganhe 70% de desconto.", # Spam.
  "Você ganhou um celular! Clique aqui para resgatar seu prêmio.", # Spam.
  "Última chance! Responda agora e concorra a um prêmio.", # Spam.
  "Oi, tudo bem? Como foi seu dia?", # Ham.
  "Vamos almoçar juntos amanhã?", # Ham.
  "Estou chegando em casa, até daqui a pouco.", # Ham.
  "Você pode me enviar aquele documento quando puder?", # Ham.
  "Bom dia! Nos encontramos às oito horas." # Ham.
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(corpus[:5])

words = vectorizer.get_feature_names_out()

words_frequency = X.toarray().sum(axis=0)

words_frequency_dictionary = [
  (words[i], words_frequency[i]) for i in range(0, len(words))
]

# Ordenando de acordo com a frequência das palavras (Decrescente).

words_frequency_dictionary.sort(key=lambda tuple: tuple[1], reverse=True)

print(f"Vocabulário:\n\n{words}")

print(f"\nMatriz:\n\n{X.toarray()}")

print("\n5 Palavras Mais Frequentes (Spam):\n")

for (word, frequency) in words_frequency_dictionary[:5]:
  print(
    f"| Palavra: {word:20}",
    f"| Frequência: {frequency:20} |"
  )

print("\nNova Frase: 'Você pode confirmar seu cadastro acessando este link?'")

print("\nConclusão: A nova frase parece ser um Ham (Comum), pois não possui as palavras mais comuns usadas em um Spam.")
```
