### 02 - Recursos (Importações)

```python
from sklearn.feature_extraction.text import CountVectorizer
```

### 03 - Resolução (Bruta)

```python
news_01 = """
A inteligência artificial está transformando diversas áreas da sociedade.
A inteligência artificial também está mudando a forma como empresas
trabalham, produzem e analisam informações. Com a inteligência artificial,
novas ferramentas podem ser desenvolvidas para auxiliar pessoas em suas
atividades.
"""

corpus_01 = [
  news_01
]

vectorizer_01 = CountVectorizer(token_pattern=r"(?u)\b\w+\b") # Permitindo palavras com 1 caractere.

X_01 = vectorizer_01.fit_transform(corpus_01)

words_01 = vectorizer_01.get_feature_names_out()

words_frequency_01 = X_01.toarray().sum(axis=0)

words_frequency_dictionary_01 = [(words_01[i], words_frequency_01[i]) for i in range(0, len(words_01))]

words_frequency_dictionary_01.sort(key=lambda tuple: tuple[1], reverse=True)

print(f"Vocabulário (Bruto):\n\n{words_01}")

print("\n10 Palavras Mais Frequentes (Bruto):\n")

for (word, frequency) in words_frequency_dictionary_01[:10]:
  print(
    f"| Palavra: {word:20}",
    f"| Frequência: {frequency:20} |"
  )
```

### 04 - Resolução (Normalizada)

```python
news_02 = """
A inteligência artificial está transformando diversas áreas da sociedade.
A inteligência artificial também está mudando a forma como empresas
trabalham, produzem e analisam informações. Com a inteligência artificial,
novas ferramentas podem ser desenvolvidas para auxiliar pessoas em suas
atividades.
"""

stop_words_list = ["e", "a", "o", "as", "os", "um", "uma", "uns", "umas", "com", "de", "da", "do", "em"]

corpus_02 = [
  news_02
]

vectorizer_02 = CountVectorizer(token_pattern=r"(?u)\b\w+\b") # Permitindo palavras com 1 caractere.

X_02 = vectorizer_02.fit_transform(corpus_02)

words_02 = vectorizer_02.get_feature_names_out()

words_frequency_02 = X_02.toarray().sum(axis=0)

words_frequency_dictionary_02 = [
  (words_02[i], words_frequency_02[i]) for i in range(0, len(words_02)) if words_02[i] not in stop_words_list
]

words_frequency_dictionary_02.sort(key=lambda tuple: tuple[1], reverse=True)

print(f"Vocabulário (Normalizado):\n\n{words_02}")

print("\n10 Palavras Mais Frequentes (Normalizado):\n")

for (word, frequency) in words_frequency_dictionary_02[:10]:
  print(
    f"| Palavra: {word:20}",
    f"| Frequência: {frequency:20} |"
  )
```
