### 03 - Exemplo Inicial

```python
nlp = spacy.load("pt_core_news_lg")

text = "O Pablo DeSeNvOlVeU o PI na Fatec."

document = nlp(text)

print("Tokens:\n")

for token in document:
  print(
    f"| Texto: {token.text:12}",
    f"| Lemma: {token.lemma_:12}",
    f"| POS: {token.pos_:12}",
    f"| Stop: {token.is_stop:12} |"
  )

print("\nEntidades:\n")

for entity in document.ents:
  print(
    f"| Texto: {entity.text:12}",
    f"| Label: {entity.label_:12} |"
  )
```
