```python
import spacy

from collections import Counter
```

### 03 - Resolução

```python
nlp_05 = spacy.load("pt_core_news_sm")

text_05 = "Na manhã de sábado, Lucas caminhou até a praça para encontrar seus amigos. O grupo conversou durante alguns minutos e decidiu organizar um pequeno passeio pela cidade. Enquanto caminhavam pelas ruas, observaram as árvores, tiraram algumas fotografias e conheceram uma nova cafeteria. Depois, todos retornaram para casa satisfeitos com o passeio e combinaram de se encontrar novamente no domingo. Foi um dia agradável."

document_05 = nlp_05(text_05)

print(f"Frase: {text_05}")

min_letters_quantity = 8 # Troque para testar.

clean_tokens = [token for token in document_05 if (not token.is_stop and not token.is_punct and len(token.text) >= min_letters_quantity)]

print(f"\nTokens (Limpos):\n")

for token in clean_tokens:
  print(
    f"| Texto: {token.text:20}",
    f"| POS: {token.pos_:20} |"
  )

print(f"\nQuantidade de Tokens (Limpos): {len(clean_tokens)}.")

clean_tokens_counter = Counter(token.text for token in clean_tokens)

print(f"\nFrequência de Aparição dos Tokens (Limpos):\n")

for key, value in clean_tokens_counter.most_common():
  token, frequency = key, value

  print(
    f"| Token: {token:20}",
    f"| Frequência: {frequency} |"
  )
```
