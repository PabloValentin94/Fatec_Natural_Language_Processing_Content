### 02 - Recursos (Importações)

```python
import spacy

from collections import Counter
```

### 03 - Resolução

```python
nlp_04 = spacy.load("pt_core_news_sm")

text_04 = "Na manhã de sábado, Lucas caminhou até a praça para encontrar seus amigos. O grupo conversou durante alguns minutos e decidiu organizar um pequeno passeio pela cidade. Enquanto caminhavam pelas ruas, observaram as árvores, tiraram algumas fotografias e conheceram uma nova cafeteria. Depois, todos retornaram para casa satisfeitos com o passeio e combinaram de se encontrar novamente no domingo."

document_04 = nlp_04(text_04)

print(f"Frase: {text_04}")

clean_tokens = [token for token in document_04 if (not token.is_stop and not token.is_punct)]

print(f"\nTokens (Limpos):\n")

for token in clean_tokens:
  print(
    f"| Texto: {token.text:20}",
    f"| POS: {token.pos_:20} |"
  )

print(f"\nQuantidade de Tokens (Limpos): {len(clean_tokens)}.")

clean_tokens_counter = Counter(token.text for token in clean_tokens)

print(f"\n5 Tokens Mais Frequentes:\n")

for key, value in clean_tokens_counter.most_common(5):
  token, frequency = key, value

  print(
    f"| Token: {token:20}",
    f"| Frequência: {frequency} |"
  )
```
