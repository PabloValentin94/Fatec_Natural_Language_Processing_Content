### 02 - Recursos (Importações)

```python
import spacy

from collections import Counter
```

### 03 - Resolução

```python
nlp_06 = spacy.load("pt_core_news_sm")

text_06 = "Na manhã de sábado, Peter Parker, Gwen Stacy, Peter Parker, Miles Morales e Peter Parker caminharam até a praça para encontrar seus amigos. O grupo conversou durante alguns minutos e decidiu organizar um pequeno passeio pela cidade. Enquanto caminhavam pelas ruas, observaram as árvores, tiraram algumas fotografias e conheceram uma nova cafeteria. Depois, todos retornaram para casa satisfeitos com o passeio e combinaram de se encontrar novamente no domingo."

document_06 = nlp_06(text_06)

print(f"Frase: {text_06}")

print("\nEntidades:\n")

for entity in document_06.ents:
  print(
    f"| Entidade: {entity.text:20}",
    f"| Tipo: {entity.label_:20} |"
  )

print(f"\nQuantidade de Entidades: {len(document_06.ents)}.")

entity_types = Counter(entity.label_ for entity in document_06.ents)

print("\nFrequência de Aparição de Tipos de Entidades:\n")

for key, value in entity_types.most_common():
  entity_type, frequency = key, value

  print(
    f"| Tipo: {entity_type:20}",
    f"| Frequência: {frequency} |"
  )

print(f"\nQuantidade de Tipos de Entidades Encontrados: {len(entity_types.keys())}.")
```
