### 03 - Resolução

```python
nlp_02 = spacy.load("pt_core_news_sm")

text_02 = "Os alunos estavam estudando algoritmos e implementaram soluções eficientes."

document_02 = nlp_02(text_02)

print(f"Frase: {text_02}")

print("\nTokens:\n")

texts, lemmas_ = [], []

for token in document_02:
  print(
    f"| Texto: {token.text:20}",
    f"| Lemma: {token.lemma_:20} |"
  )

  texts.append(token.text)
  lemmas_.append(token.lemma_)

print(f"\nTextos: {texts}")
print(f"\nLemmas: {lemmas_}")

print("\nComparações:\n")

for i in range(0, len(texts)):
  text_document = nlp_02(texts[i])

  lemma_document = nlp_02(lemmas_[i])

  similarity = text_document.similarity(lemma_document)

  print(
    f"| Texto: {text_document.text:20}",
    f"| Lemma: {lemma_document.text:20}",
    f"| Similaridade: {(similarity * 100):20.2f}% |"
  )
```
