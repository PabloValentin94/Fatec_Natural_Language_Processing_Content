### 03 - Resolução

```python
nlp_01 = spacy.load("pt_core_news_sm")

text_01 = "A inteligência artificial está transformando a educação superior no Brasil."

document_01 = nlp_01(text_01)

print(f"Frase: {text_01}")

print("\nTokens:\n")

for token in document_01:
  print(
    f"| Texto: {token.text:20}",
    f"| POS: {token.pos_:20} |"
  )

print(f"\nQuantidade de Tokens: {len(document_01)}")

print("\nTokens (Palavras):\n")

word_tokens = [item for item in document_01 if (item.pos_ not in ["SYM", "PUNCT", "X"])]

for token in word_tokens:
  print(
    f"| Texto: {token.text:20}",
    f"| POS: {token.pos_:20} |"
  )

print(f"\nQuantidade de Tokens (Palavras): {len(word_tokens)}")
```
