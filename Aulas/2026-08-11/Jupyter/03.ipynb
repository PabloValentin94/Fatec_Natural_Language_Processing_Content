### 03 - Resolução

```python
nlp_03 = spacy.load("pt_core_news_sm")

text_03 = "Na manhã de sábado, Lucas caminhou até a praça para encontrar seus amigos. O grupo conversou durante alguns minutos e decidiu organizar um pequeno passeio pela cidade. Enquanto caminhavam pelas ruas, observaram as árvores, tiraram algumas fotografias e conheceram uma nova cafeteria. Depois, todos retornaram para casa satisfeitos com o passeio e combinaram de se encontrar novamente no domingo."

document_03 = nlp_03(text_03)

print(f"Frase: {text_03}")

print("\nTokens:\n")

for token in document_03:
  print(
    f"| Texto: {token.text:20}",
    f"| POS: {token.pos_:20} |"
  )

print(f"\nQuantidade de Tokens: {len(document_03)}.")

verbs = [verb for verb in document_03 if (verb.pos_ == "VERB")]

print("\nTokens (Verbos):\n")

for token in verbs:
  print(
    f"| Texto: {token.text:20}",
    f"| POS: {token.pos_:20} |"
  )

print(f"\nQuantidade de Tokens (Verbos): {len(verbs)}.")

nouns = [noun for noun in document_03 if (noun.pos_ == "NOUN")]

print("\nTokens (Substantivos):\n")

for token in nouns:
  print(
    f"| Texto: {token.text:20}",
    f"| POS: {token.pos_:20} |"
  )

print(f"\nQuantidade de Tokens (Substantivos): {len(nouns)}.")

print("\nSujeitos (Verbos):\n")

for token in verbs:
  verb_subjects = [children.text for children in token.children if (children.dep_ == "nsubj")]

  subject_description = ', '.join(verb_subjects) if (len(verb_subjects) > 0) else 'Não há sujeito'

  print(
    f"| Verbo: {token.text:20}",
    f"| POS: {token.pos_:20}",
    f"| Sujeito(s): {subject_description:20} |"
  )
```
