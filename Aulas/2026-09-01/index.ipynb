### 02 - Recursos (Importações)

```python
from gensim.models import Word2Vec
```

### 03 - Exemplo Inicial

```python
sentences = [
  ["gato", "come", "peixe"],
  ["cachorro", "come", "carne"],
  ["gato", "e", "cachorro", "são", "animais"]
]

model = Word2Vec(sentences, vector_size=5, window=2, min_count=1)

print(model.wv['gato'])

print(f"\n{model.wv['cachorro']}")
```
