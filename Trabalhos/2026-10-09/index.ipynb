### 02 - Recursos (Importações)

```python
from typing import Literal

import pandas as pd
```

### 03 - Classe

```python
class Bag_Words:
  def __init__(self, corpus: list[str], frequency_mode: Literal["binary", "count"] = "count"):
    self.documents: list[str] = corpus

    self.frequency_mode = frequency_mode

    self.vocabulary: list[str] = []

    self.occurrence_matrix: list[list[int]] = []

  def __generate_occurrence_matrix(self):
    matrix = []

    for document in self.documents:
      document_tokens = document.split()

      if (self.frequency_mode == "binary"): # Verifica somente se existe (Sim/Não <-> 1/0).
        matrix.append([
          (1 if token in document_tokens else 0) for token in self.vocabulary
        ])
      else: # Conta a quantidade de vezes que o token aparece.
        matrix.append([
          (document_tokens.count(token) if token in document_tokens else 0) for token in self.vocabulary
        ])

    return matrix

  def fit_transform(self):
    for document in self.documents:
      document_tokens = document.split()

      self.vocabulary.extend(
        token for token in document_tokens
          if (len(token) > 1 and token not in self.vocabulary) # O vocabulário terá tokens únicos e com mais de um caractere.
      )

      self.vocabulary.sort(reverse=False) # Organiza os tokens em ordem crescente dentro da lista original.

    self.occurrence_matrix = self.__generate_occurrence_matrix()

    return self.occurrence_matrix

  def get_features_names_out(self):
    return self.vocabulary
```

### 04 - Implementação

```python
corpus = [
  "gosto de machine learning e gosto de programação",
  "machine learning é incrível",
  "gosto de programar"
]

bag_words = Bag_Words(corpus)

X = bag_words.fit_transform()

print(
  "Vocabulário:",
  "\n\n",
  bag_words.get_features_names_out()
)

print(
  "\nMatriz:",
  "\n\n",
  X
)

print(
  "\nDataFrame:",
  "\n\n",
  pd.DataFrame(
    data=X,
    columns=bag_words.get_features_names_out(),
    index=[f"Documento {i + 1}" for i in range(0, len(corpus))]
  ),
)
```
