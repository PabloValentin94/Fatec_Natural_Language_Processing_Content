### 02 - Recursos (Importações)

```python
import spacy
import requests
from bs4 import BeautifulSoup
```

### 03 - Extração de Notícias Oficiais

```python
site_url = 'https://www.globo.com/'

response = requests.get(site_url)

html_content = response.text

soup = BeautifulSoup(html_content, 'html.parser')

news_titles = soup.find_all('h2', class_='post__title')

print("Títulos das Notícias:\n")

for title in news_titles:
    print(f"| {title.text.strip():120} |")
```

### 04 - Extração de Tokens

```python
nlp = spacy.load("pt_core_news_lg")

news = " ".join(news.text.strip() for news in news_titles)

document = nlp(news)

print("Tokens:\n")

for token in document:
  print(
    f"| Texto: {token.text:20}",
    f"| Lemma: {token.lemma_:20}",
    f"| POS (Pos Tagging): {token.pos_:10}",
    f"| Stop Word: {'Sim' if (token.is_stop == 1) else 'Não'} ({int(token.is_stop)}) |"
  )
```

### 05 - Extração de Entidades

```python
print("Entidades:\n")

for entity in document.ents:
  print(
    f"| Texto: {entity.text:30}",
    f"| Label: {entity.label_:10} |"
  )
```

### 06 - Filtragem de Entidades

```python
search = [entity for entity in document.ents if (entity.label_ in ['ORG', 'MISC', 'PER', 'DATE'])]

print("\nPessoas, Organizações e Datas:\n")

for entity in search:
  print(
    f"| Texto: {entity.text:30}",
    f"| Label: {entity.label_:10} |"
  )
```
